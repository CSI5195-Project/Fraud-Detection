```python
import pandas as pd

configs = [
    'P2-S20',
    'P20-S20',
    'P200-S20',
    'P2-S50',
    'P20-S50',
    'P200-S50', 
]
```

```python
trials = pd.DataFrame()

for config in configs:
    df = pd.read_csv(f"/kaggle/input/model2-optimizations/study_CIARP2024-{config}_results.csv")
    df['population'] = int(config.split('P')[1].split('-')[0])
    df['step'] = int(config.split('S')[1].split('-')[0])
    df['P_S'] = [config + "_" + str(num) for num in df['number']]
    trials = pd.concat([trials, df])

trials
```

```python
Q1 = trials['user_attrs_@5FPR aod_age'].quantile(0.25)  # 25th percentile
Q3 = trials['user_attrs_@5FPR aod_age'].quantile(0.75)  # 75th percentile
IQR = Q3 - Q1

threshold = Q3 + 1.5 * IQR
threshold
```

```python
best_trial = trials[trials['user_attrs_@5FPR aod_age'] <= threshold]

best_trial_sorted = best_trial.sort_values(by='user_attrs_@5FPR recall', ascending=False)
best_trial_selected = best_trial_sorted.head(1)
best_trial_selected
```

# Reformatting hyperparams

```python
# removing params
best_trial_selected.rename(
    columns={col: col.replace("params_", "") for col in best_trial_selected.columns if col.startswith("params_")},
    inplace=True
)
```

```python
# combining betas
beta_columns = [col for col in best_trial_selected.columns if col.startswith("beta")]
best_trial_selected["beta"] = best_trial_selected[beta_columns].apply(tuple, axis=1)

# best_trial_selected = best_trial_selected.drop(columns=beta_columns)
```

```python
# combining adam_betas
abeta_columns = [col for col in best_trial_selected.columns if col.startswith("adam_beta")]
best_trial_selected["adam_beta"] = best_trial_selected[abeta_columns].apply(tuple, axis=1)

# best_trial_selected = best_trial_selected.drop(columns=abeta_columns)
```

```python
# combining thesholds
thresh_columns = [col for col in best_trial_selected.columns if col.startswith("threshold")]
best_trial_selected["threshold"] = best_trial_selected[thresh_columns].apply(tuple, axis=1)

# best_trial_selected = best_trial_selected.drop(columns=thresh_columns)
```

```python
best_trial_selected["batch"] = [1024]
best_trial_selected["epoch"] = [5]
```
